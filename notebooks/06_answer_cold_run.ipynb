### 🔧 Change applied on 2026-10-24: fix 1, whole-day periods (half-open intervals)

**What was wrong.** The half-years were selected with `slot_at.between("2025-01-01", "2025-06-30")`. A date written without a time is read as **midnight**, so the upper bound was `2025-06-30 00:00:00`, and every booking on **30 June after midnight was left out** of both halves. Notebook 04's SQL uses `slot_at < '2025-07-01'`, which keeps the whole day, so pandas and SQL disagreed:

| | pandas before the fix | SQL, and pandas after the fix | Bookings on 30 June that were dropped |
|---|---|---|---|
| H1 2025 tests | 431,261 | **433,611** | 1,032 bookings |
| H1 2026 tests | 452,933 | **455,383** | 1,015 bookings |
| Growth | +5.03% | **+5.02%** | |

The growth rate barely moves, which is why the bug is easy to miss. The **levels** were wrong, though, and the levels are what finance audits and what has to match the SQL.

**What changed.** A small helper, `in_period(ts, period)`, selects `ts >= start` and `ts < end + 1 day`. CONFIG still states the period as the readable `('2026-01-01', '2026-06-30')`, and the helper turns it into a half-open interval, so no period anywhere can lose its last day again.

**Effect on this notebook.** Every count over the two halves now includes 30 June, and every number below agrees with notebook 04's SQL.

```python
bk = pd.read_csv(CONFIG["clean"] / "bookings_clean.csv.gz", parse_dates=["slot_at"], dtype={"booking_id": "string", "person_id": "string", "rescheduled_from": "string", "status_at_export": "string"})
# CHANGE (fix 1, 2026-10-24): whole-day half-open periods, so 30 June is counted (it was dropped by between()).
def in_period(ts, period):
    """Whole days, half-open: on or after the first day, before the day after the last day.
    ('2026-01-01', '2026-06-30') keeps every timestamp from 1 Jan 00:00 to 30 Jun 23:59:59."""
    start = pd.Timestamp(period[0])
    end_exclusive = pd.Timestamp(period[1]) + pd.Timedelta(days=1)
    return (ts >= start) & (ts < end_exclusive)
h = {"last": bk[in_period(bk["slot_at"], CONFIG["period_last"])], "this": bk[in_period(bk["slot_at"], CONFIG["period_this"])]}
tests_last, tests_this = int(h["last"]["n_tests"].sum()), int(h["this"]["n_tests"].sum())
growth = (tests_this / tests_last - 1) * 100
HEADLINE = {"claim": "<one sentence for the board>", "number": round(growth, 1), "numerator": tests_this, "denominator": tests_last,
            "unit": "tests on bookings that were collected or no-show, after the identity rule in docs/decisions-log.md",
            "period": "H1 2026 against H1 2025, six cities", "comparison": "plan: +18 percent", "caveat": "<the one thing that would change the claim>"}
for k, v in HEADLINE.items(): print(f"{k:>12}: {v}")
```
